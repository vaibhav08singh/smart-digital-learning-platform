````python
def generate_quiz(pdf_file, client, num_questions=10):

    prompt = f"""
You are an educational quiz generator.

Read the uploaded document carefully and create exactly {num_questions}
multiple-choice questions based ONLY on the document.

For every question provide:

- question
- options A, B, C, D
- correct_answer
- topic
- difficulty

Rules:

- Use ONLY information present in the document.
- Do not invent facts.
- Each question must have exactly one correct answer.
- Cover different topics from the document.
- Return ONLY valid JSON.
- Do not use markdown code fences.

Required JSON format:

{{
  "questions": [
    {{
      "question": "Question text",
      "options": {{
        "A": "Option A",
        "B": "Option B",
        "C": "Option C",
        "D": "Option D"
      }},
      "correct_answer": "A",
      "topic": "Topic name",
      "difficulty": "Easy"
    }}
  ]
}}
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=[
            pdf_file,
            prompt
        ]
    )

    text = response.text.strip()

    # Remove markdown fences if Gemini adds them
    if text.startswith("```"):
        text = text.replace("```json", "", 1)
        text = text.replace("```", "", 1)
        text = text.strip()

    quiz_data = json.loads(text)

    # Validate generated quiz
    validation_result = validate_generated_quiz(
        quiz_data,
        expected_questions=num_questions
    )

    if not validation_result["valid"]:
        return {
            "success": False,
            "quiz_data": quiz_data,
            "validation_errors": validation_result["errors"]
        }

    return {
        "success": True,
        "quiz_data": quiz_data,
        "validation_errors": []
    }


print("generate_quiz function updated with validation successfully!")
````

````python
from collections import defaultdict

def analyze_quiz(quiz_data, student_answers):

    results = []

    # 1. Check individual answers
    for i, q in enumerate(quiz_data["questions"]):

        correct = q["correct_answer"]
        student = student_answers[i]

        result = "Correct" if student == correct else "Wrong"

        results.append({
            "question": q["question"],
            "student_answer": student,
            "correct_answer": correct,
            "result": result,
            "topic": q["topic"]
        })

    # 2. Topic-wise performance
    topic_stats = defaultdict(lambda: {
        "correct": 0,
        "total": 0
    })

    for i, q in enumerate(quiz_data["questions"]):

        topic = q["topic"]

        topic_stats[topic]["total"] += 1

        if student_answers[i] == q["correct_answer"]:
            topic_stats[topic]["correct"] += 1

    topic_performance = {}

    for topic, stats in topic_stats.items():

        accuracy = (
            stats["correct"] / stats["total"]
        ) * 100

        topic_performance[topic] = {
            "correct": stats["correct"],
            "total": stats["total"],
            "accuracy": round(accuracy, 2)
        }

    # 3. Identify knowledge gaps
    knowledge_gaps = []

    for topic, stats in topic_performance.items():

        if stats["accuracy"] < 60:
            knowledge_gaps.append(topic)

    return {
        "results": results,
        "topic_performance": topic_performance,
        "knowledge_gaps": knowledge_gaps
    }


print("analyze_quiz function created successfully!")
````

````python
def validate_answers(quiz_data, student_answers):

    total_questions = len(quiz_data["questions"])
    total_answers = len(student_answers)

    if total_questions != total_answers:
        return False, (
            f"Expected {total_questions} answers, "
            f"but received {total_answers}."
        )

    for answer in student_answers:
        if answer not in ["A", "B", "C", "D"]:
            return False, f"Invalid answer found: {answer}"

    return True, "Answers are valid."


print("validate_answers function created successfully!")
````

````python
def calculate_score(quiz_data, student_answers):

    correct_answers = [
        question["correct_answer"]
        for question in quiz_data["questions"]
    ]

    score = 0

    for student_answer, correct_answer in zip(
        student_answers, correct_answers
    ):
        if student_answer == correct_answer:
            score += 1

    total_questions = len(correct_answers)

    percentage = (score / total_questions) * 100

    return {
        "score": score,
        "total_questions": total_questions,
        "percentage": percentage
    }


print("calculate_score function created successfully!")
````

````python
def generate_quiz_analysis(quiz_data, student_answers):

    # Validate answers first
    is_valid, validation_message = validate_answers(
        quiz_data,
        student_answers
    )

    if not is_valid:
        return {
            "success": False,
            "message": validation_message
        }

    # Calculate score
    score_result = calculate_score(
        quiz_data,
        student_answers
    )

    # Question-wise analysis
    question_analysis = analyze_answers(
        quiz_data,
        student_answers
    )

    correct_count = score_result["score"]
    total_questions = score_result["total_questions"]
    wrong_count = total_questions - correct_count

    return {
        "success": True,
        "score": correct_count,
        "total_questions": total_questions,
        "correct_answers": correct_count,
        "wrong_answers": wrong_count,
        "percentage": score_result["percentage"],
        "question_analysis": question_analysis
    }


print("generate_quiz_analysis function created successfully!")
````

````python
def generate_performance_insights(final_analysis):

    if not final_analysis["success"]:
        return {
            "success": False,
            "message": final_analysis["message"]
        }

    percentage = final_analysis["percentage"]

    if percentage >= 80:
        performance = "Excellent"
        message = "Great performance! Keep up the good work."

    elif percentage >= 60:
        performance = "Good"
        message = "Good performance, but there is still room for improvement."

    elif percentage >= 40:
        performance = "Needs Improvement"
        message = "You have a basic understanding, but some topics need more practice."

    else:
        performance = "Needs Significant Improvement"
        message = "More practice and revision are recommended."

    return {
        "success": True,
        "performance": performance,
        "message": message,
        "percentage": percentage
    }


print("generate_performance_insights function created successfully!")
````

````python
performance_insights = generate_performance_insights(
    final_analysis
)

print("Performance Insights:")
print(performance_insights)
````

````python
def identify_weak_questions(final_analysis):

    if not final_analysis["success"]:
        return {
            "success": False,
            "message": final_analysis["message"]
        }

    weak_questions = []

    for result in final_analysis["question_analysis"]:

        if not result["is_correct"]:
            weak_questions.append({
                "question_number": result["question_number"],
                "student_answer": result["student_answer"],
                "correct_answer": result["correct_answer"]
            })

    return {
        "success": True,
        "weak_questions": weak_questions,
        "weak_count": len(weak_questions)
    }


print("identify_weak_questions function created successfully!")
````

````python
def identify_weak_topics(quiz_data, final_analysis):

    if not final_analysis["success"]:
        return {
            "success": False,
            "message": final_analysis["message"]
        }

    weak_topics = {}

    for result in final_analysis["question_analysis"]:

        if not result["is_correct"]:

            question_number = result["question_number"]

            question = quiz_data["questions"][question_number - 1]

            topic = question.get("topic", "Unknown")

            if topic in weak_topics:
                weak_topics[topic] += 1
            else:
                weak_topics[topic] = 1

    return {
        "success": True,
        "weak_topics": weak_topics
    }


print("identify_weak_topics function created successfully!")
````

````python
weak_topic_result = identify_weak_topics(
    quiz_data,
    final_analysis
)

print("Weak Topic Analysis:")
print(weak_topic_result)
````

````python
def rank_weak_topics(weak_topic_result):

    if not weak_topic_result["success"]:
        return {
            "success": False,
            "message": weak_topic_result["message"]
        }

    weak_topics = weak_topic_result["weak_topics"]

    ranked_topics = sorted(
        weak_topics.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return {
        "success": True,
        "ranked_weak_topics": ranked_topics
    }


print("rank_weak_topics function created successfully!")
````

````python
ranked_topic_result = rank_weak_topics(
    weak_topic_result
)

print("Ranked Weak Topics:")
print(ranked_topic_result)
````

````python
def analyze_difficulty_performance(quiz_data, final_analysis):

    if not final_analysis["success"]:
        return {
            "success": False,
            "message": final_analysis["message"]
        }

    difficulty_stats = {}

    for result in final_analysis["question_analysis"]:

        question_number = result["question_number"]

        question = quiz_data["questions"][question_number - 1]

        difficulty = question.get("difficulty", "Unknown")

        if difficulty not in difficulty_stats:
            difficulty_stats[difficulty] = {
                "total": 0,
                "correct": 0,
                "wrong": 0
            }

        difficulty_stats[difficulty]["total"] += 1

        if result["is_correct"]:
            difficulty_stats[difficulty]["correct"] += 1
        else:
            difficulty_stats[difficulty]["wrong"] += 1

    # Calculate percentage for each difficulty
    for difficulty, stats in difficulty_stats.items():

        stats["percentage"] = (
            stats["correct"] / stats["total"]
        ) * 100

    return {
        "success": True,
        "difficulty_performance": difficulty_stats
    }


print("analyze_difficulty_performance function created successfully!")
````

````python
difficulty_result = analyze_difficulty_performance(
    quiz_data,
    final_analysis
)

print("Difficulty-wise Performance:")
print(difficulty_result)
````

````python
def generate_student_report(
    final_analysis,
    performance_insights,
    weak_question_result,
    ranked_topic_result,
    difficulty_result
):

    if not final_analysis["success"]:
        return {
            "success": False,
            "message": final_analysis["message"]
        }

    return {
        "success": True,

        "summary": {
            "score": final_analysis["score"],
            "total_questions": final_analysis["total_questions"],
            "correct_answers": final_analysis["correct_answers"],
            "wrong_answers": final_analysis["wrong_answers"],
            "percentage": final_analysis["percentage"],
            "performance": performance_insights["performance"]
        },

        "weak_questions": weak_question_result["weak_questions"],

        "weak_topics": ranked_topic_result["ranked_weak_topics"],

        "difficulty_performance": difficulty_result[
            "difficulty_performance"
        ],

        "question_analysis": final_analysis["question_analysis"]
    }


print("generate_student_report function created successfully!")
````

````python
def create_student_summary(student_report):

    if not student_report["success"]:
        return "Unable to generate student report."

    summary = student_report["summary"]

    score = summary["score"]
    total = summary["total_questions"]
    percentage = summary["percentage"]
    performance = summary["performance"]

    ranked_topics = student_report["weak_topics"]

    if ranked_topics:
        top_topic = ranked_topics[0][0]
    else:
        top_topic = "No major weak topic identified"

    difficulty_data = student_report["difficulty_performance"]

    lowest_difficulty = None
    lowest_percentage = 101

    for difficulty, data in difficulty_data.items():

        if data["percentage"] < lowest_percentage:
            lowest_percentage = data["percentage"]
            lowest_difficulty = difficulty

    summary_text = (
        f"You scored {score}/{total} ({percentage:.2f}%). "
        f"Overall performance: {performance}. "
        f"Your main weak topic is '{top_topic}'. "
        f"You performed lowest on {lowest_difficulty}-level questions "
        f"with {lowest_percentage:.2f}% accuracy."
    )

    return summary_text


print("create_student_summary function created successfully!")
````

````python
def generate_recommendation(student_report):

    if not student_report["success"]:
        return {
            "success": False,
            "message": student_report["message"]
        }

    ranked_topics = student_report["weak_topics"]

    difficulty_data = student_report["difficulty_performance"]

    recommendations = []

    # Recommend weak topics
    if ranked_topics:

        top_topic = ranked_topics[0][0]

        recommendations.append(
            f"Revise the topic: {top_topic}."
        )

    # Recommend weakest difficulty level
    weakest_difficulty = None
    lowest_percentage = 101

    for difficulty, data in difficulty_data.items():

        if data["percentage"] < lowest_percentage:
            lowest_percentage = data["percentage"]
            weakest_difficulty = difficulty

    if weakest_difficulty:

        recommendations.append(
            f"Practice more {weakest_difficulty}-level questions."
        )

    # General recommendation
    recommendations.append(
        "Review incorrect questions and attempt a similar quiz again."
    )

    return {
        "success": True,
        "recommendations": recommendations
    }


print("generate_recommendation function created successfully!")
````

````python
def evaluate_quiz(quiz_data, student_answers):

    # Step 1: Generate complete rule-based analysis
    final_analysis = generate_quiz_analysis(
        quiz_data,
        student_answers
    )

    if not final_analysis["success"]:
        return final_analysis

    # Step 2: Performance insights
    performance_insights = generate_performance_insights(
        final_analysis
    )

    # Step 3: Weak questions
    weak_question_result = identify_weak_questions(
        final_analysis
    )

    # Step 4: Weak topics
    weak_topic_result = identify_weak_topics(
        quiz_data,
        final_analysis
    )

    # Step 5: Rank weak topics
    ranked_topic_result = rank_weak_topics(
        weak_topic_result
    )

    # Step 6: Difficulty performance
    difficulty_result = analyze_difficulty_performance(
        quiz_data,
        final_analysis
    )

    # Step 7: Student report
    student_report = generate_student_report(
        final_analysis,
        performance_insights,
        weak_question_result,
        ranked_topic_result,
        difficulty_result
    )

    # Step 8: Human-readable summary
    student_summary = create_student_summary(
        student_report
    )

    # Step 9: Rule-based recommendations
    recommendation_result = generate_recommendation(
        student_report
    )

    # Step 10: Create final evaluation
    final_evaluation = create_final_evaluation_response(
        student_report,
        student_summary,
        recommendation_result
    )

    # Step 11: Prepare data for Gemini
    ai_feedback_data = prepare_ai_feedback_data(
        final_evaluation
    )

   # Step 12: Prepare detailed weak-question data
    weak_question_details = prepare_weak_question_details(
        quiz_data,
        final_analysis
    )
   # Step 13: Create question-aware Gemini prompt
    ai_prompt = create_ai_feedback_prompt(
        ai_feedback_data,
        weak_question_details
    ) 
    

    # Step 14: Generate safe AI feedback
    ai_result = generate_ai_feedback(
        ai_prompt
    )

    # Step 15: Add AI feedback
    if ai_result["success"]:

        final_evaluation["ai_feedback"] = ai_result["feedback"]
        final_evaluation["ai_feedback_status"] = "generated"

    else:

        final_evaluation["ai_feedback"] = (
            "AI feedback is temporarily unavailable. "
            "Please use the performance summary and recommendations."
        )

        final_evaluation["ai_feedback_status"] = "unavailable"

    return final_evaluation


print("evaluate_quiz updated with AI feedback successfully!")
````

````python
def validate_generated_quiz(quiz_data, expected_questions=10):

    errors = []

    # 1. Check questions key
    if "questions" not in quiz_data:
        return {
            "success": False,
            "valid": False,
            "errors": ["Missing 'questions' key."]
        }

    questions = quiz_data["questions"]

    # 2. Check number of questions
    if len(questions) != expected_questions:
        errors.append(
            f"Expected {expected_questions} questions, "
            f"but got {len(questions)}."
        )

    # 3. Validate each question
    for i, question in enumerate(questions, start=1):

        required_fields = [
            "question",
            "options",
            "correct_answer"
        ]

        for field in required_fields:
            if field not in question:
                errors.append(
                    f"Question {i}: missing '{field}'."
                )

        # Stop checking this question if required fields are missing
        if not all(field in question for field in required_fields):
            continue

        # Check options
        options = question["options"]

        if not isinstance(options, dict):
            errors.append(
                f"Question {i}: options must be a dictionary."
            )
            continue

        required_options = ["A", "B", "C", "D"]

        for option in required_options:
            if option not in options:
                errors.append(
                    f"Question {i}: missing option '{option}'."
                )

        # Check correct answer
        if question["correct_answer"] not in required_options:
            errors.append(
                f"Question {i}: invalid correct answer "
                f"'{question['correct_answer']}'."
            )

    # 4. Final result
    if errors:
        return {
            "success": True,
            "valid": False,
            "errors": errors
        }

    return {
        "success": True,
        "valid": True,
        "errors": []
    }


print("Quiz validation function created successfully!")
````
